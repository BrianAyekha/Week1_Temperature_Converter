## 1. Function and threshold

```python
# Function that converts a temperature from Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

# Threshold in °C that the temperature is compared against
# (37.0 °C is roughly normal body temperature, relevant to infant warmers)
THRESHOLD_C = 37.0

# Function that reports whether a temperature is above, below, or at the threshold
def check_threshold(celsius, threshold=THRESHOLD_C):
    if celsius > threshold:
        return f"above the threshold of {threshold} °C"
    elif celsius < threshold:
        return f"below the threshold of {threshold} °C"
    else:
        return f"equal to the threshold of {threshold} °C"
```

## 2. Convert one temperature entered by the user

```python
# Ask the user for a temperature in °C (float() turns the text into a number)
try:
    temp_c = float(input("Enter a temperature in °C: "))
    temp_f = celsius_to_fahrenheit(temp_c)   # call the conversion function
    print(f"{temp_c} °C = {temp_f:.1f} °F")
    print("This temperature is", check_threshold(temp_c))
except ValueError:
    # Runs if the user types something that is not a number
    print("Please enter a valid number.")
```

## 3. Convert several temperatures using a loop

```python
# A list of temperatures (°C) to convert
temperatures_c = [34.5, 36.0, 37.0, 38.2, 40.1]

# Loop through each temperature, convert it, and check it against the threshold
for t in temperatures_c:
    f = celsius_to_fahrenheit(t)
    print(f"{t:5.1f} °C = {f:5.1f} °F -> {check_threshold(t)}")
```

## 4. (Optional) Keep converting until the user types 'q'

```python
# A while loop repeats until the user chooses to stop
while True:
    entry = input("Enter a temperature in °C (or 'q' to quit): ")
    if entry.lower() == "q":
        print("Done.")
        break
    try:
        t = float(entry)
        print(f"{t} °C = {celsius_to_fahrenheit(t):.1f} °F -> {check_threshold(t)}")
    except ValueError:
        print("Please enter a valid number.")
```
